# Распознавание цифры на защите

Ноутбук загружает веса выбранной по валидации модели из `models/best_model.npz`. Для проверки нужен файл с одной цифрой на однотонном фоне. Запустите ноутбук из каталога `Lab6` с установленными зависимостями из `requirements.txt` и пакетом `notebook`.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
from PIL import Image, ImageOps

lab_dir = Path.cwd()
src_dir = str(lab_dir / "src")
if src_dir not in sys.path:
    sys.path.insert(0, src_dir)

from inference import predict_image, preprocess_image
from network import load_model

model = load_model(lab_dir / "models" / "best_model.npz")
```

## Изображение для проверки

Подставьте путь к новому файлу в ячейке ниже и запустите её вместе со следующей ячейкой. Для пробы доступны `data/drawings/2.png`, `data/drawings/5.png` и `data/drawings/7.png`.

```python
image_path = lab_dir / "data" / "drawings" / "2.png"
# Для произвольного файла используйте Path("C:/path/to/digit.png").
```

```python
with Image.open(image_path) as original:
    original = ImageOps.exif_transpose(original).convert("RGBA")
prepared = preprocess_image(image_path)
prediction, probabilities = predict_image(model, image_path)

fig, axes = plt.subplots(1, 2, figsize=(7, 3))
axes[0].imshow(original)
axes[0].set_title("Исходный рисунок")
axes[1].imshow(prepared, cmap="gray", vmin=0, vmax=1, interpolation="nearest")
axes[1].set_title("Вход модели 28 × 28")
for ax in axes:
    ax.axis("off")
plt.tight_layout()
plt.show()

print(f"Предсказанная цифра — {prediction}")
print("Цифра | Вероятность")
for digit, probability in enumerate(probabilities):
    print(f"{digit:>5} | {probability:>11.4%}")
```
